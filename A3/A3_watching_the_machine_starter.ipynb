```python
import pandas as pd

ages = ["20", " 21 ", "", "N/A", "twenty", "22", "20"]

# 1. Convert to a Series and strip leading/trailing whitespace
s = pd.Series(ages).str.strip()

# 2. Convert to numeric; unparseable values become NaN
s_clean = pd.to_numeric(s, errors="coerce")

# 3. Calculate average (pandas automatically ignores NaN values)
average_age = s_clean.mean()

print("Cleaned Series:")
print(s_clean)
print(f"\nAverage age: {average_age:.2f}")
```
