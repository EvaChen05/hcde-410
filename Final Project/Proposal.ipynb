# Final Project Proposal: Trust and Reliance on AI Advice

**Course:** HCDE 410

## Overview & research questions / hypotheses
### Overview
This study will examine how self-reported trust in AI advice relates to changes from correct answers to incorrect recommendations. I will combine quantitative analysis of a public dataset with one interview exploring an individual’s decisions about AI advice.

### Research question
Is higher self-reported trust in AI advice associated with a higher rate of switching from a correct answer to an incorrect AI recommendation?

### Hypotheses
Participants who report higher trust in AI advice will have a higher rate of switching from correct answers to incorrect AI recommendations.

## Background / Related Work
### Background
The background of this project can be traced to the concept of trust calibration, which means that users’ trust in AI should match its actual capabilities (Muir, 1987; Lee & See, 2004). This concept motivates my study of whether people who report higher trust in AI are more likely to change correct answers to match incorrect AI advice.

### Related work
Lee and See (2004) describe trust calibration as the match between users’ trust and a system’s capabilities. Their work emphasizes appropriate reliance on automation. This motivates my study of both reported trust and users’ actual decisions.

Buçinca et al. (2020) found that subjective trust and preference did not predict objective decision-making performance in their experiments. This suggests that trust ratings alone may not show whether users rely on AI appropriately. My study will compare trust ratings with correct-to-incorrect answer changes, using the advice and before-and-after responses collected by Vodrahalli et al. (2022).

Buçinca et al. (2021) found that interaction designs encouraging careful thinking reduced overreliance, but also involved trade-offs in users’ evaluations of the designs. This informs my interview questions about what makes users check AI advice and what makes checking difficult. The interview will explore one person’s experience alongside the quantitative analysis.

## Methodology

### Quantitative Analysis

I will analyze records labeled as AI advice. For each participant within each task, I will calculate how often they changed a correct answer to match incorrect advice. I will check missing values and repeated participant records.
I will use descriptive statistics and correlation analysis to examine the relationship between trust ratings and switching rates.I will analyze tasks separately and report the direction, strength, uncertainty, and statistical significance of the relationships. The specific correlation method will be selected after checking the data.
The switching rate will be the number of correct-to-incorrect changes divided by the number of valid cases where the participant initially answered correctly and received incorrect advice. Participants with no eligible cases will not be assigned a rate of zero.

### Qualitative Interview

I will conduct one 30–45 minute semi-structured interview with a volunteer who has used AI advice. I will ask about specific experiences of accepting, checking, and rejecting advice. I will code the interview notes and group related ideas to understand the participant’s decision process.
I will obtain consent and share only anonymous findings. The interview findings will describe one person’s experience and will not be treated as representative of all AI users.

### Presenting Findings

I will present the quantitative findings using summary tables and scatterplots. I will compare these findings with insights from the interview, while recognizing that the interview participant is separate from the dataset participants. The study will examine associations.

## Motivation & Problem
1. AI can give useful advice, but it can also make mistakes. Users need to decide when to accept or reject its advice.
2. I want to find out whether people who report higher trust in AI accept its incorrect advice more often.
3. I hope to understand whether higher trust is linked to overreliance on AI, meaning that users follow its advice even when it is wrong.

## Data

**Dataset:** I will use the Human–AI Interactions Dataset by Vodrahalli and colleagues. It contains over 30,000 interactions across five classification tasks.

**Source:** [Human–AI Interactions Dataset](https://github.com/kailas-v/human-ai-interactions)

**Contents:** The dataset includes users’ trust ratings, the advice they received, and their answers before and after seeing the advice. It also shows whether the advice was correct or incorrect.

**Trust measurement:** Trust ratings were collected after all questions in an experiment, rather than after each question. I will group responses by participant within each task.

**Fit with my question:** I will select records where `advice_source == "ai"`, meaning the advice was presented as coming from AI. This label does not necessarily mean an AI model generated the 
advice. I will define accepting incorrect advice as changing an initially correct answer to match an incorrect recommendation. I will compare users’ trust ratings with how often these changes occur.

**License:** The repository provides an MIT License. I will cite the original study and keep the required copyright and license notice when reusing its materials.

**Ethical considerations:** I will report group-level results and will not try to identify participants. The data come from specific experimental tasks, so the findings may not apply to everyday AI use.
