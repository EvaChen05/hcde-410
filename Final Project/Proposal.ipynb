# Final Project Proposal: Trust and Reliance on AI Advice

**Course:** HCDE 410

## Research Question
Do users who report higher trust in AI accept incorrect AI advice significantly more often?

## Motivation & Problem
1. AI can give useful advice, but it can also make mistakes. Users need to decide when to accept or reject its advice.
2. I want to find out whether people who report higher trust in AI accept its incorrect advice more often.
3. I hope to understand whether higher trust is linked to overreliance on AI, meaning that users follow its advice even when it is wrong.

## Data
**Dataset:** I will use the Human–AI Interactions Dataset by Vodrahalli and colleagues. It contains over 30,000 interactions across five classification tasks.

**Source:** https://github.com/kailas-v/human-ai-interactions

**Contents:** The dataset includes users’ trust ratings, the advice they received, and their answers before and after seeing the advice. It also shows whether the advice was correct or incorrect.

**Fit with my question:** I will use records where the advice was presented as coming from AI. I will compare users’ trust ratings with how often their final answers follow incorrect AI advice.

**License:** The repository provides an MIT License. I will cite the original study and keep the required copyright and license notice when reusing its materials.

**Ethical considerations:** I will report group-level results and will not try to identify participants. The data come from specific experimental tasks, so the findings may not apply to everyday AI use.
